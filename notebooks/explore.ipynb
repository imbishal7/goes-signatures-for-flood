```python

import geopandas as gpd
import pandas as pd
```

```python
df = gpd.read_parquet('../data/raw/groundsource_2026.parquet')
df['start_date'] = pd.to_datetime(df['start_date'])
df['end_date'] = pd.to_datetime(df['end_date'])
```

```python
df_conus = df.cx[-125:-66, 24:49]
```

```python
df = df_conus[df_conus['start_date']>='2020'].copy()
```

```python
total_area_cov = df['area_km2'].sum()
total_area_cov
```

```python
def plot_areas(gdf, color_by="start_date", n=100000):
    return gdf.sample(n).explore(
        column=color_by,
        tooltip=["start_date", "end_date", "area_km2"],
        popup=True,
        cmap="viridis",
        style_kwds={"fillOpacity": 0.4},
    )


plot_areas(df)
```
