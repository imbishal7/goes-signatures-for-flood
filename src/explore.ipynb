```python
def plot_areas(gdf, color_by="start_date", n=100000):
    return gdf.sample(n).explore(
        column=color_by,
        tooltip=["start_date", "end_date", "area_km2"],
        popup=True,
        cmap="viridis",
        style_kwds={"fillOpacity": 0.4},
    )


plot_areas(df)
```
